```python
import matplotlib.pyplot as plt
import cv2
import numpy as np
from itertools import product

# Define junction rules
JUNCTION_TABLE = {
    'L': [('C', 'O'), ('O', 'C')],
    'T': [('C', 'O', 'O'), ('O', 'C', 'O')],
    'Y': [('C', 'C', 'C')],
    'Arrow': [('C', 'C', 'O'), ('C', 'O', 'C')],
}

EDGE_LABELS = ['C', 'V', 'O']  # Concave, Convex, Occluding

# Edge class
class Edge:
    def __init__(self, id, coords):
        self.id = id
        self.coords = coords  # (x1, y1, x2, y2)
        self.possible_labels = EDGE_LABELS.copy()

# Junction class
class Junction:
    def __init__(self, id, edges):
        self.id = id
        self.edges = edges

    def set_type(self, junction_type):
        self.type = junction_type

    def constrain(self):
        combos = list(product(*[e.possible_labels for e in self.edges]))
        valid = [c for c in combos if c in JUNCTION_TABLE.get(self.type, [])]
        new_possible = [set() for _ in self.edges]
        for combo in valid:
            for i, label in enumerate(combo):
                new_possible[i].add(label)
        for i, edge in enumerate(self.edges):
            edge.possible_labels = list(set(edge.possible_labels) & new_possible[i])

# Waltz interpreter
class WaltzInterpreter:
    def __init__(self, edges, junctions):
        self.edges = edges
        self.junctions = junctions

    def run(self):
        updated = True
        while updated:
            updated = False
            for junc in self.junctions:
                before = [e.possible_labels.copy() for e in junc.edges]
                junc.constrain()
                after = [e.possible_labels for e in junc.edges]
                if before != after:
                    updated = True

# Detect edges using OpenCV
def detect_edges(image_path):
    img = cv2.imread(image_path)
    if img is None:
        print("Error: Image not found.")
        return [], None

    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    canny = cv2.Canny(gray, 50, 150)
    lines = cv2.HoughLinesP(canny, 1, np.pi/180, threshold=80, minLineLength=30, maxLineGap=10)

    edge_list = []
    if lines is not None:
        for idx, line in enumerate(lines):
            x1, y1, x2, y2 = line[0]
            edge_list.append(Edge(idx, (x1, y1, x2, y2)))
    return edge_list, img

# Draw with enhanced visuals
def draw_output(img, edges):
    fig, ax = plt.subplots(figsize=(12, 10))
    ax.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    ax.set_title("Waltz Line Labeling - Enhanced Visualization", fontsize=16)
    ax.axis("off")

    color_map = {'C': 'red', 'V': 'green', 'O': 'blue'}

    for edge in edges:
        x1, y1, x2, y2 = edge.coords
        midx, midy = (x1 + x2) // 2, (y1 + y2) // 2

        # Draw directional arrow
        ax.annotate("",
                    xy=(x2, y2), xytext=(x1, y1),
                    arrowprops=dict(arrowstyle="->", color='blue', lw=1.8))

        # Label edge ID + labels with visual offset
        label_text = f"E{edge.id}\n" + "/".join(edge.possible_labels)
        ax.text(midx + 8, midy - 8, label_text,
                fontsize=9, color='darkgreen',
                bbox=dict(facecolor='white', edgecolor='black', boxstyle='round,pad=0.4'))

    plt.tight_layout()
    plt.show()

# Main script
if __name__ == "__main__":
    IMAGE_PATH = "/content/ex.png"  # Replace with actual image path

    edges, image = detect_edges(IMAGE_PATH)

    if len(edges) >= 3:
        # Dummy junctions – replace with actual detection if needed
        j1 = Junction("J1", [edges[0], edges[1]])
        j1.set_type('L')

        j2 = Junction("J2", [edges[1], edges[2]])
        j2.set_type('L')

        interpreter = WaltzInterpreter(edges, [j1, j2])
        interpreter.run()

        draw_output(image, edges)
    else:
        print("Not enough edges detected to form junctions.")
```

import cv2
import numpy as np
import matplotlib.pyplot as plt
from itertools import product

# --- STEP 1: Define Valid Junction Rules ---
JUNCTION_TABLE = {
    'L': [('C', 'O'), ('O', 'C')],
    'T': [('C', 'O', 'O'), ('O', 'C', 'O')],
    'Y': [('C', 'C', 'C')],
    'Arrow': [('C', 'C', 'O'), ('C', 'O', 'C')],
}

EDGE_LABELS = ['C', 'V', 'O']  # Concave, Convex, Occluding

# --- STEP 2: Define Edge and Junction Classes ---
class Edge:
    def __init__(self, id, coords):
        self.id = id
        self.coords = coords  # (x1, y1, x2, y2)
        self.possible_labels = EDGE_LABELS.copy()

class Junction:
    def __init__(self, id, edges, coords):
        self.id = id
        self.edges = edges
        self.coords = coords  # (x, y) position for label display
        self.type = None
        self.inferred_label = None

    def set_type(self, junction_type):
        self.type = junction_type

    def constrain(self):
        label_combinations = list(product(*[e.possible_labels for e in self.edges]))
        valid = []
        for combo in label_combinations:
            if combo in JUNCTION_TABLE.get(self.type, []):
                valid.append(combo)

        new_possible = [set() for _ in self.edges]
        for combo in valid:
            for i, label in enumerate(combo):
                new_possible[i].add(label)

        for i, edge in enumerate(self.edges):
            edge.possible_labels = list(set(edge.possible_labels) & new_possible[i])

    def infer_label(self):
        if self.type in JUNCTION_TABLE:
            labels = [e.possible_labels for e in self.edges]
            for combo in product(*labels):
                if combo in JUNCTION_TABLE[self.type]:
                    if 'C' in combo:
                        self.inferred_label = 'Concave'
                    elif 'V' in combo:
                        self.inferred_label = 'Convex'
                    else:
                        self.inferred_label = 'Occluding'
                    return
        self.inferred_label = 'Unknown'

# --- STEP 3: Waltz Interpreter ---
class WaltzInterpreter:
    def __init__(self, edges, junctions):
        self.edges = edges
        self.junctions = junctions

    def run(self):
        updated = True
        while updated:
            updated = False
            for junction in self.junctions:
                before = [e.possible_labels.copy() for e in junction.edges]
                junction.constrain()
                after = [e.possible_labels for e in junction.edges]
                if before != after:
                    updated = True

    def print_results(self):
        for edge in self.edges:
            print(f"Edge {edge.id}: Coordinates {edge.coords} -> Possible Labels {edge.possible_labels}")

# --- STEP 4: Image Edge Detection ---
def detect_edges_from_image(image_path):
    img = cv2.imread(image_path)
    if img is None:
        print(f"Error: Could not load image from {image_path}.")
        return [], None

    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    edges = cv2.Canny(gray, 50, 150, apertureSize=3)
    lines = cv2.HoughLinesP(edges, 1, np.pi/180, threshold=80, minLineLength=30, maxLineGap=10)

    edge_objects = []
    if lines is not None:
        for idx, line in enumerate(lines):
            x1, y1, x2, y2 = line[0]
            edge_objects.append(Edge(idx, (x1, y1, x2, y2)))
            cv2.line(img, (x1, y1), (x2, y2), (0, 0, 255), 2)
    return edge_objects, img

# --- STEP 5: Draw Junction Labels on Image ---
def draw_junction_labels(img, junctions):
    for junction in junctions:
        if junction.inferred_label:
            x, y = junction.coords
            cv2.putText(
                img,
                junction.inferred_label,
                (x + 5, y - 5),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.5,
                (0, 255, 0),
                1,
                cv2.LINE_AA
            )
    return img

# --- STEP 6: Main ---
if __name__ == "__main__":
    image_path = "r/content/pic.jpg.png  # Replace with your image path
    edges, img = detect_edges_from_image(image_path)

    if img is not None and len(edges) >= 3:
        # Define junctions manually with coordinates (x, y)
        junction1 = Junction('J1', [edges[0], edges[1]], coords=(100, 100))
        junction1.set_type('L')
        junction1.infer_label()

        junction2 = Junction('J2', [edges[1], edges[2]], coords=(150, 150))
        junction2.set_type('L')
        junction2.infer_label()

        interpreter = WaltzInterpreter(edges, [junction1, junction2])
        interpreter.run()
        interpreter.print_results()

        labeled_img = draw_junction_labels(img, [junction1, junction2])
        plt.imshow(cv2.cvtColor(labeled_img, cv2.COLOR_BGR2RGB))
        plt.title("Labeled Junctions")
        plt.axis("off")
        plt.show()
    else:
        print("Not enough edges detected or image failed to load.")
